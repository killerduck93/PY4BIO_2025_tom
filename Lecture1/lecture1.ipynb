### print formatting I

```python
print("My name is {}.".format("Luka"))
myName="Khan"
print("My name is {}.".format(myName))
```

### print formatting II

```python
# left align
print("Yes, {:10s} is my name.".format("Luka"))
print("Yes, {:<10s} is my name.".format("Luka"))
```

```python
myName="Khan"
# left align
print("Yes, {:<10s} is my name.".format(myName))
# center align
print("Yes, {:^10s} is my name.".format(myName))
# right align
print("Yes, {:>10s} is my name.".format(myName))
```

### Enumerate

```python
i = 0
for nuc in "actgc":
    print("%i: %s" % (i+1, nuc))
    i += 1
```

```python
for i, nuc in enumerate("actgc"):
    print("%i: %s" % (i+1, nuc))
```
