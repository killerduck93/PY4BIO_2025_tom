## 1. Reading Sequence Files with SeqIO

`SeqIO` is Biopython's sequence input/output interface. It can read and write various sequence file formats.

### Reading FASTA Files
FASTA is one of the most common sequence file formats. Let's parse a multi-FASTA file:

```python
from Bio import SeqIO
for seq_record in SeqIO.parse("multi_fasta_gb_header.fasta", "fasta"):
    print( seq_record.id )
    print( repr(seq_record.seq) )
    #print( seq_record.seq )
    print( len(seq_record) )
```

**Note:** 
- `seq_record.id` - gives the sequence identifier
- `seq_record.seq` - gives the actual sequence
- `len(seq_record)` - gives the sequence length

### Reading GenBank Files
GenBank format contains more annotation information than FASTA:

```python
for seq_record in SeqIO.parse("sequence.gb", "genbank"):
    print(seq_record)
    #added to print just one record example
    break 
```

Let's extract specific information from each GenBank record:

```python
for seq_record in SeqIO.parse("sequence.gb", "genbank"):
    print(seq_record.id)
    print(repr(seq_record.seq))
    print(len(seq_record))
```

### Reading Frames
Extract different reading frames from a DNA sequence:

```python
my_seq = Seq("GATCGATGGGCCTATATAGGATCGAAAATCGC")
```

```python
my_seq[0::3]
```

```python
my_seq[1::3]
```

```python
my_seq[2::3]
```

**Explanation of slicing:**
- `my_seq[0::3]` - starts at position 0, takes every 3rd nucleotide (reading frame 1)
- `my_seq[1::3]` - starts at position 1, takes every 3rd nucleotide (reading frame 2)
- `my_seq[2::3]` - starts at position 2, takes every 3rd nucleotide (reading frame 3)

## 9. BLAST - Sequence Similarity Searching

### Running BLAST Online
Biopython can submit BLAST queries to NCBI's web service:

Draw an ASCII representation of the phylogenetic tree:

```python
from Bio.Blast import NCBIWWW
```

```python
result_handle = NCBIWWW.qblast("blastn", "nt", "8332116")
```

**Parameters:**
- `blastn` - nucleotide BLAST
- `nt` - nucleotide database
- `8332116` - GI number to search

**Note:** This can take several minutes to run!

### Parsing BLAST Results

```python
from Bio.Blast import NCBIXML
```

```python
blast_record = NCBIXML.read(result_handle)
```

### Parsing BLAST XML Output with E-value Filtering
Filter BLAST results by E-value to find significant matches:

```python
E_VALUE_THRESH = 1e-20
for record in NCBIXML.parse(open("insulin.xml")):
    if record.alignments:
        print("\n")
        print("query: %s" % record.query[:100])
        for align in record.alignments:
            for hsp in align.hsps:
                if hsp.expect < E_VALUE_THRESH:
                    print("match: %s " % align.title[:100])
```

**E-value (Expect value):** Lower values indicate more significant matches. An E-value of 1e-20 means the match is very unlikely to occur by chance.
