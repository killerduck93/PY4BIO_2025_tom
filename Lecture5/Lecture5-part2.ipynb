### Biopython motif objects

# Introduction to Biopython Motifs

This notebook introduces **sequence motifs** and how to work with them using Biopython's `motifs` module.

## What is a Sequence Motif?

A **motif** is a recurring pattern in DNA, RNA, or protein sequences that often has biological significance. For example:
- **Transcription factor binding sites** in DNA
- **Promoter elements** that regulate gene expression
- **Conserved protein domains**

In bioinformatics, we analyze motifs to:
1. Identify regulatory elements in genomes
2. Predict protein binding sites
3. Find conserved functional sequences across species

Biopython provides powerful tools to create, analyze, and search for motifs in biological sequences.

```python
from Bio import motifs
```

```python
from Bio.Seq import Seq
```

## Creating a Motif from Sequence Instances

To create a motif in Biopython, we need:
1. A collection of **aligned sequences** (instances) that share a common pattern
2. Each instance should be a `Seq` object from Biopython

Let's create a motif from a list of 5-nucleotide sequences that represent potential binding sites.

```python
instances = [Seq("TACAA"),Seq("TACGC"),Seq("TACAC"),Seq("TACCC"),Seq("AACCC"),Seq("AATGC"),Seq("AATGC")]
```

```python
instances
```

### Viewing the Instances

Our instances are the individual sequences we'll use to build the motif. Notice they all have the same length (5 bp) and share some common nucleotides at certain positions.

```python
m = motifs.create(instances)
```

### Creating the Motif Object

The `motifs.create()` function analyzes our sequences and creates a motif object. This object contains:
- **Counts matrix**: How many times each nucleotide (A, C, G, T) appears at each position
- **Consensus sequence**: The most common nucleotide at each position
- Other useful representations (PWM, PSSM) we'll explore later

```python
print(m)
```

### Visualizing the Motif

Printing the motif displays all the instances in an alignment format. This helps us visually identify conserved positions.

```python
print(m.counts)
```

### Understanding the Counts Matrix

The **counts matrix** (also called frequency matrix) shows how many times each nucleotide appears at each position:
- Each column represents a position in the motif (positions 0-4)
- Each row represents a nucleotide (A, C, G, T)
- The values tell us the frequency of each nucleotide at each position

This matrix is fundamental for motif analysis and scoring sequences.

```python
m.consensus
```

### The Consensus Sequence

The **consensus sequence** represents the most common nucleotide at each position. It's a simplified representation of the motif, useful for:
- Quick visual identification
- Pattern matching
- Comparing motifs

The consensus is automatically calculated from the counts matrix.

```python
arnt = motifs.read(open("MA0004.1.sites"), "sites")
```

## Reading Motifs from Files

In real-world bioinformatics, motifs are often stored in standard file formats. Biopython can read various formats:
- **sites**: Simple text file with aligned sequences
- **pfm**: Position Frequency Matrix
- **jaspar**: JASPAR database format
- **meme**: MEME suite output

Here we read a motif for the ARNT (Aryl Hydrocarbon Receptor Nuclear Translocator) transcription factor from a sites file.

```python
print(arnt.counts)
```

### Examining the ARNT Motif

Notice that this motif has many more instances than our simple example. The counts are higher because it's derived from actual experimental data.

```python
print(arnt.instances[:3])
```

### Viewing Individual Instances

We can access the original sequences that make up the motif. Here we look at the first 3 instances to see what actual binding sites look like.

```python
for instance in arnt.instances:
    print(instance)
```

### Iterating Through All Instances

We can loop through all instances to examine or process each binding site sequence individually.

```python
with open("SRF.pfm") as handle:
    srf = motifs.read(handle, "pfm")
```

## Working with Position Frequency Matrices (PFM)

A **PFM** (Position Frequency Matrix) is another way to represent motifs. It contains the same information as a counts matrix but may come from different sources or tools.

Here we read a PFM for the SRF (Serum Response Factor) transcription factor. Unlike the sites format which contains sequences, a PFM directly contains the count values.

```python
srf
```

### The SRF Motif Object

When created from a PFM file, the motif object still has the same properties as before.

```python
print(srf.counts)
```

```python
print(srf.instances)
```

### Important Note: PFM vs Sites

When a motif is created from a PFM file (just the matrix), the `instances` attribute will be empty because we don't have the original sequences - only their summarized counts!

```python
print(arnt.counts.consensus)
```

### Comparing Consensus Sequences

Let's compare the consensus sequences of different transcription factors. This helps us understand their binding preferences.

```python
print(srf.counts.consensus)
```

```python
print(srf.pwm)
```

## Advanced Motif Representations

### Position Weight Matrix (PWM)

The **PWM** converts counts to **frequencies** (probabilities). For each position:
- Each value = count of that nucleotide / total sequences
- Values range from 0 to 1
- All four nucleotides at a position sum to 1

PWMs are useful for:
- Comparing motifs of different sample sizes
- Probabilistic scoring of sequences

```python
print(srf.pssm)
```

### Position-Specific Scoring Matrix (PSSM)

The **PSSM** (also called log-odds matrix) is calculated from the PWM:
- Compares the motif frequency to background frequency
- Uses logarithms for easier score calculation
- Positive values: nucleotide is more common than background
- Negative values: nucleotide is less common than background

PSSMs are the standard way to **score sequences** for motif matches!

```python
print(srf.pwm.log_odds())
```

### Alternative Way to Get PSSM

You can also get the log-odds (PSSM) directly from the PWM using the `log_odds()` method. This is equivalent to accessing `pssm` directly.

# MEME

## Parsing MEME Output

**MEME** (Multiple EM for Motif Elicitation) is a widely-used tool for discovering motifs in biological sequences. It:
- Automatically finds motifs in a set of sequences
- Doesn't require you to know the motif beforehand
- Provides statistical significance (E-values)

Biopython can parse MEME output files, allowing you to programmatically access discovered motifs and their properties.

### Reading MEME Results

MEME typically outputs results in XML format, which Biopython can parse to extract motif information.

```python
handle = open("meme_run_2.xml")
```

```python
record = motifs.parse(handle, "meme")
```

```python
handle.close()
```

### Understanding the MEME Record

The `record` object is a container for all motifs found by MEME. Each motif has:
- A consensus sequence
- Occurrence information
- Statistical significance
- Position-specific scoring information

```python
len(record)
```

### How Many Motifs Did MEME Find?

MEME often finds multiple motifs in your input sequences. Let's see how many it discovered.

```python
motif = record[0]
```

### Accessing Individual Motifs

We can access motifs by index (like a list). The first motif (index 0) is typically the most significant one found by MEME.

```python
print(motif.consensus)
```

### Consensus and Degenerate Consensus

**Consensus**: The single most common nucleotide at each position.

**Degenerate consensus**: Uses IUPAC ambiguity codes when multiple nucleotides are common at a position:
- **R** = A or G (puRine)
- **Y** = C or T (pYrimidine)
- **W** = A or T (Weak)
- **S** = G or C (Strong)
- **K** = G or T (Keto)
- **M** = A or C (aMino)
- **N** = any nucleotide

The degenerate consensus captures more information about variability in the motif.

```python
print(motif.degenerate_consensus)
```

```python
motif.num_occurrences
```

### Motif Properties

MEME provides rich metadata about each discovered motif. Let's explore some key properties:

```python
motif.length
```

```python
evalue = motif.evalue
```

### Statistical Significance: E-value

The **E-value** (Expectation value) tells us how likely it is to find this motif by chance:
- **Lower E-values** = more significant motifs
- E-value < 0.05 is typically considered significant
- E-value << 0.001 is highly significant

MEME uses sophisticated statistical models to calculate these values.

```python
print("%3.1g" % evalue)
```

```python
motif.name
```

### Motif Name

MEME assigns each motif a name, usually based on its consensus or degenerate consensus sequence.

```python
motif = record['TGTGKSTG']
```

### Accessing Motifs by Name

Instead of using numeric indices, you can access motifs by their name (consensus sequence). This is useful when you know which motif you're looking for.

```python
len(motif.instances)
```

## Working with Motif Instances from MEME

MEME not only finds the motif pattern but also reports **where** it found each occurrence in your input sequences. Let's explore these instances.

```python
motif.instances[0]
```

### Examining Individual Instances

Each instance represents one occurrence of the motif. Let's look at the first one in detail.

```python
motif.instances[0].start
```

### Instance Position Information

Each instance has metadata about its location:
- **start**: The position where the motif begins in the original sequence (0-indexed)
- **strand**: Which DNA strand ('+' or '-') the motif was found on
- **length**: The length of the motif
- **pvalue**: Statistical significance of this particular occurrence

```python
motif.instances[0].strand
```

```python
motif.instances[0].length
```

```python
pvalue = motif.instances[0].pvalue
```

### Instance P-value

The **p-value** for each instance tells us how well this particular sequence matches the motif:
- Lower p-values indicate better matches
- This helps distinguish strong binding sites from weak ones

```python
print("%5.3g" % pvalue)
```
