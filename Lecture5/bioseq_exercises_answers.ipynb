#### Exercise 1: Creating and Manipulating Sequences with **Bio.Seq**
1. Create a DNA sequence using Bio.Seq.
2. Transcribe the DNA sequence to RNA.
3. Translate the RNA sequence to a protein sequence.
4. Find the reverse complement of the DNA sequence.
5. Calculate the GC content of the DNA sequence.

```python
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction

# 1. Create a DNA sequence
dna_seq = Seq("ATGCGTACGTAGCTAGCTAGCTAGCTAGCTAGC")

# 2. Transcribe the DNA sequence to RNA
rna_seq = dna_seq.transcribe()
print("RNA Sequence:", rna_seq)

# 3. Translate the RNA sequence to a protein sequence
protein_seq = rna_seq.translate()
print("Protein Sequence:", protein_seq)

# 4. Find the reverse complement of the DNA sequence
rev_comp_seq = dna_seq.reverse_complement()
print("Reverse Complement DNA Sequence:", rev_comp_seq)

# 5. Calculate the GC content of the DNA sequence
gc_content = gc_fraction(dna_seq) * 100
print(f'GC Content: {gc_content:.2f}%')
```

#### Exercise 2: Reading and Writing Sequence Files with Bio.SeqIO
1. Read a FASTA file containing multiple sequences ("E2F_up_promoters.fa").
2. Print the ID and sequence of each record.
3. Write the sequences to a new FASTA file.
4. Convert the FASTA file to a GenBank file.
5. Count the number of sequences in the FASTA file.

```python
from Bio import SeqIO

# 1. Read a FASTA file
fasta_file = "E2F_up_promoters.fa"
records = list(SeqIO.parse(fasta_file, "fasta"))

# 2. Print the ID and sequence of each record
for record in records:
    print(f"ID: {record.id}")
    print(f"Sequence: {record.seq}\n")

# 3. Write the sequences to a new FASTA file
output_fasta = "output.fasta"
SeqIO.write(records, output_fasta, "fasta")

# 4. Add molecule_type annotation and convert the FASTA file to a GenBank file
for record in records:
    if "molecule_type" not in record.annotations:
        record.annotations["molecule_type"] = "DNA"

output_genbank = "output.gb"
SeqIO.write(records, output_genbank, "genbank")

# 5. Count the number of sequences in the FASTA file
num_sequences = len(records)
print(f"Number of sequences: {num_sequences}")
```

#### Exercise 3: Extracting Sub-sequences and Features
1. Read a GenBank file ("sequence.gb").
2. Extract and print the sequence of a specific gene ("5.8S rRNA").
3. Extract and print the sequence of the first CDS (coding sequence) feature.
4. Extract and print the sequence of the first exon feature.
5. Calculate the GC content of the extracted gene sequence.

```python
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction

# 1. Read a GenBank file
genbank_file = "sequence.gb"
record = SeqIO.read(genbank_file, "genbank")

# 2. Extract and print the sequence of a specific gene
gene_name = "5.8S rRNA"
for feature in record.features:
    if feature.type == "gene" and "gene" in feature.qualifiers and feature.qualifiers["gene"][0] == gene_name:
        gene_seq = feature.extract(record.seq)
        print(f"Gene Sequence ({gene_name}):", gene_seq)

# 3. Extract and print the sequence of the first CDS feature
for feature in record.features:
    if feature.type == "CDS":
        cds_seq = feature.extract(record.seq)
        print("First CDS Sequence:", cds_seq)
        break

# 4. Extract and print the sequence of the first exon feature
for feature in record.features:
    if feature.type == "exon":
        exon_seq = feature.extract(record.seq)
        print("First Exon Sequence:", exon_seq)
        break

# 5. Calculate the GC content of the extracted gene sequence
gc_content = gc_fraction(gene_seq) * 100
print(f'GC Content of {gene_name}: {gc_content:.2f}%')
```

#### Exercise 4: Sequence Alignment
1. Create two DNA sequences.
2. Perform a pairwise alignment of the sequences using **Bio.pairwise2**.
3. Print the alignment results.
4. Calculate the alignment score.
5. Identify the regions of similarity between the sequences.

```python
from Bio import pairwise2
from Bio.Seq import Seq

# 1. Create two DNA sequences
seq1 = Seq("ATGCGTACGTAGCTAGCTAGCTAGCTAGCTAGC")
seq2 = Seq("ATGCGTACGTTGCTAGCTAGCTAGCTAGCTAGC")

# 2. Perform a pairwise alignment of the sequences
alignments = pairwise2.align.globalxx(seq1, seq2)

# 3. Print the alignment results
for alignment in alignments:
    print(pairwise2.format_alignment(*alignment))

# 4. Calculate the alignment score
alignment_score = alignments[0][2]
print(f"Alignment Score: {alignment_score}")

# 5. Identify the regions of similarity between the sequences
similarity_regions = [alignment for alignment in alignments if alignment[2] == alignment_score]
print("Regions of Similarity:")
for region in similarity_regions:
    print(pairwise2.format_alignment(*region))
```

#### Exercise 5: Parsing and Analyzing Sequence Annotations
1. Read a GenBank file ("sequence.gb").
2. Print the annotations of the sequence.
3. Extract and print the organism name.
4. Extract and print the sequence length.
5. Extract and print the list of features (e.g., genes, CDS, exons).


```python
from Bio import SeqIO

# 1. Read a GenBank file
genbank_file = "sequence.gb"
record = SeqIO.read(genbank_file, "genbank")

# 2. Print the annotations of the sequence
print("Annotations:", record.annotations)

# 3. Extract and print the organism name
organism = record.annotations.get("organism", "Unknown")
print("Organism:", organism)

# 4. Extract and print the sequence length
sequence_length = len(record.seq)
print("Sequence Length:", sequence_length)

# 5. Extract and print the list of features
features = record.features
print("Features:")
for feature in features:
    print(f"Type: {feature.type}, Location: {feature.location}")
```
