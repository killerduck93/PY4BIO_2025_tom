**Exercise 17 — Mini bioinformatics task: nucleotide counts**  
Write a function `nt_counts(seq)` that returns a dictionary with counts of `'A'`, `'C'`, `'G'`, `'T'` in an uppercase DNA sequence. Ignore any other characters.  
Example: `nt_counts("ACGTTAacgNX")` → `{'A': 2, 'C': 2, 'G': 2, 'T': 3}`


````python
# Your work here
````

<details>
<summary><strong>View answer</strong></summary>

```python
def nt_counts(seq: str):
    counts = {'A': 0, 'C': 0, 'G': 0, 'T': 0}
    for ch in seq.upper():
        if ch in counts:
            counts[ch] += 1
    return counts
```
Uppercase the input and count only `A/C/G/T`. Other characters are ignored.


</details>

**Exercise 18 — Dictionary comprehension**


</details>
