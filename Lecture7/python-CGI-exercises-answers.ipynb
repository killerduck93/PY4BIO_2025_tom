## Exercise 1: Introduction to CGI

**Question:**  
What is CGI (Common Gateway Interface), and how does it work in the context of web servers and Python scripts?

**Answer:**  
CGI stands for Common Gateway Interface. It is a standard protocol that allows web servers to execute external programs, often scripts, and send their output to web clients. In the context of Python, CGI scripts are Python programs that run on the server side to generate dynamic web content. When a user requests a CGI script, the web server executes the script, which can process data, interact with databases, and produce HTML output.

## Exercise 3: Handling GET Requests

**Question:**  
Create a Python CGI script that reads query parameters from a GET request and displays them back to the user.

**Answer:**  

```python
#!/usr/bin/env python3

import cgi

print("Content-Type: text/html\n")

form = cgi.FieldStorage()
params = {key: form.getvalue(key) for key in form.keys()}

print("<html>")
print("<head><title>GET Parameters</title></head>")
print("<body>")
print("<h1>Received GET Parameters:</h1>")
print("<ul>")
for key, value in params.items():
    print(f"<li>{key}: {value}</li>")
print("</ul>")
print("</body>")
print("</html>")
```

## Exercise 4: Handling POST Requests

**Question:**  
Write a Python CGI script that processes form data submitted via POST method and displays a personalized greeting.

**Answer:**  

```python
#!/usr/bin/env python3

import cgi, cgitb
cgitb.enable()

print("Content-Type: text/html\n")

form = cgi.FieldStorage()
name = form.getvalue('name')

print("<html>")
print("<head><title>POST Data</title></head>")
print("<body>")
if name:
    print(f"<h1>Hello, {name}!</h1>")
else:
    print("<h1>No name provided.</h1>")
print("</body>")
print("</html>")
```

## Exercise 5: Processing Form Data

**Question:**  
Develop a Python CGI script that calculates the sum of two numbers provided by the user through an HTML form.

**Answer:**  

```python
#!/usr/bin/env python3

import cgi

print("Content-Type: text/html\n")

form = cgi.FieldStorage()
num1 = form.getvalue('num1')
num2 = form.getvalue('num2')

print("<html>")
print("<head><title>Sum Calculator</title></head>")
print("<body>")
try:
    total = float(num1) + float(num2)
    print(f"<h1>The sum of {num1} and {num2} is {total}.</h1>")
except (TypeError, ValueError):
    print("<h1>Error: Please provide two numbers.</h1>")
print("</body>")
print("</html>")
```

## Exercise 6: Working with Cookies

**Question:**  
Create a Python CGI script that sets a cookie with the user's favorite color and retrieves it on subsequent visits.

**Answer:**  

```python
#!/usr/bin/env python3

import os
import http.cookies as Cookie
import cgi

form = cgi.FieldStorage()
color = form.getvalue('color')

cookie = Cookie.SimpleCookie(os.environ.get('HTTP_COOKIE'))
stored_color = cookie.get('color')

print("Content-Type: text/html")

if color:
    print(f"Set-Cookie: color={color}; path=/")
    stored_color = color

print("\n")

print("<html>")
print("<head><title>Favorite Color</title></head>")
print("<body>")
if stored_color:
    print(f"<h1>Your favorite color is {stored_color.value}!</h1>")
else:
    print("<h1>Please select your favorite color.</h1>")
    print("""
    <form method="post" action="">
        <input type="text" name="color">
        <input type="submit" value="Submit">
    </form>
    """)
print("</body>")
print("</html>")
```

## Exercise 7: Error Handling in CGI Scripts

**Question:**  
Modify a Python CGI script to handle exceptions gracefully and display an error message instead of a server error.

**Answer:**  

```python
#!/usr/bin/env python3

import cgi, cgitb
cgitb.enable(display=0, logdir="/path/to/logdir")  # Disable browser error display

print("Content-Type: text/html\n")

print("<html>")
print("<head><title>Error Handling</title></head>")
print("<body>")
try:
    # Example operation that may cause an error
    result = 10 / 0
    print(f"<h1>Result is {result}</h1>")
except Exception as e:
    print("<h1>An error occurred while processing your request.</h1>")
print("</body>")
print("</html>")
```

## Exercise 8: File Upload Handling

**Question:**  
Write a Python CGI script that allows users to upload a file and saves it on the server.

**Answer:**  

```python
#!/usr/bin/env python3

import cgi, os

form = cgi.FieldStorage()
fileitem = form['file']

print("Content-Type: text/html\n")
print("<html>")
print("<head><title>File Upload</title></head>")
print("<body>")

if fileitem.filename:
    fn = os.path.basename(fileitem.filename)
    open('/path/to/uploads/' + fn, 'wb').write(fileitem.file.read())
    print(f"<h1>File '{fn}' uploaded successfully.</h1>")
else:
    print("<h1>No file was uploaded.</h1>")

print("</body>")
print("</html>")
```

## Exercise 10: Using Environment Variables

**Question:**  
Write a Python CGI script that displays all the environment variables available to the CGI script.

**Answer:**  

```python
#!/usr/bin/env python3

import os

print("Content-Type: text/html\n")
print("<html>")
print("<head><title>Environment Variables</title></head>")
print("<body>")
print("<h1>Environment Variables</h1>")
print("<ul>")
for key, value in os.environ.items():
    print(f"<li>{key}: {value}</li>")
print("</ul>")
print("</body>")
print("</html>")
```
