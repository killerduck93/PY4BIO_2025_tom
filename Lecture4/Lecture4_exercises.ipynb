records = {}
cur = None
for line in data.splitlines():
    line = line.strip()
    if not line: continue
    if line.startswith('>'):
        cur = line[1:]
        records[cur] = []
    else:

<details>
<summary><strong>View answer</strong></summary>

```python
class Motif:
    def __init__(self, name: str, pattern: str, flags=0):
        self.name = name
        self.pattern = re.compile(pattern, flags)
    def find(self, seq: str):
        return [(m.start(), m.end()) for m in self.pattern.finditer(seq)]
```
